## Data Visualization

Create visualizations to better understand the relationships in the data.

### Heatmap: Correlation Between Features

A heatmap shows the correlation between numeric variables in the dataset. Correlation values range from -1 to +1:
- **+1:** Perfect positive correlation (when one increases, the other increases)
- **-1:** Perfect negative correlation (when one increases, the other decreases)
- **0:** No correlation

### Stacked Bar Chart: Survival by Cabin Class

A stacked bar chart shows the number of survivors and non-survivors in each cabin class.
   - What pattern do you observe? Which class had the most survivors?

The code structure is provided. Run the cell to generate the plots.

```python
# Create a figure with two subplots
fig, axes = plt.subplots(1, 2, figsize=(15, 6))

# Plot 1: Heatmap of correlations
numeric_cols = ['Survived', 'Pclass', 'Age', 'SibSp', 'Parch', 'Fare']
correlation_matrix = df[numeric_cols].corr()

sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f',
            linewidths=0.5, ax=axes[0], cbar_kws={'label': 'Correlation'})
axes[0].set_title('Feature Correlation Heatmap', fontsize=12, fontweight='bold')

# Plot 2: Number of survivors and non-survivors by cabin class (Stacked bar chart)
survived_by_class = df.groupby(['Pclass', 'Survived']).size().unstack()
survived_by_class.plot(kind='bar', stacked=True, ax=axes[1],
                       color=['red', 'green'], edgecolor='black')
axes[1].set_title('Number of Passengers by Cabin Class', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Cabin Class')
axes[1].set_ylabel('Number of Passengers')
axes[1].legend(title='Survival Status', labels=['Deceased', 'Survived'])

for i, (idx, row) in enumerate(survived_by_class.iterrows()):
    axes[1].text(i, row[0]/2, f'{int(row[0])}', ha='center', va='center', color='white', fontweight='bold')
    axes[1].text(i, row[0] + row[1]/2, f'{int(row[1])}', ha='center', va='center', color='white', fontweight='bold')

plt.tight_layout()
plt.show()
```
